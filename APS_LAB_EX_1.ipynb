1.A fair coin is tossed 100 times.Calculate:
•	Number of heads
•	Number of tails
•	Experimental probability of Head
•	Experimental probability of Tai


```python
import random

n=100

tosses = [random.choice(["Head", "Tail"]) for _ in range(n)]

heads= tosses.count("Head")
tails= tosses.count("Tail")

probability_head= heads/n
probability_tail= tails/n

print("Number of Heads:", heads)
print("Number of Tails:", tails)

print("Experimental Probability of Head:", probability_head)
print("Experimental Probability of Tail:", probability_tail)
```

2. check Effect of the number of trials
Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000


```python
import random

trail_values=[10,100,1000,10000]

print("Fair coin toss Experiment for Number of Trails\n")

for n in trail_values:

    print(f"\nNumber of Trials: {n}")

    tosses=[random.choice(["Head","Tail"]) for _ in range(n)]

    heads=tosses.count("Head")
    tails=tosses.count("Tail")

    probability_head=heads/n
    probability_tail=tails/n

    print("Number of Heads:",heads)
    print("Number of Tails:",tails)

    print("Experimental Probability of Head:",probability_head)
    print("Experimental Probability of Tail:",probability_tail)

```

3. Rolling a die: A fair die is rolled 6,000 times.
Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

```python
import random

n=6000

rolls=[random.randint(1,6) for _ in range(n)]

print("DIE ROLLING EXPERIMENT FOR 6000 TRAIL \n")

for outcome in range(1,7):
    frequency=rolls.count(outcome)

    probability=frequency/n

    print(f"Outcome {outcome}")
    print(f"Frequency: {frequency}")
    print(f"Probability: {probability}\n\n")
```

4.Addition rule for disjoint events
A die is rolled 10,000 times.
Define: A=getting 1, B= getting 6, check the events are joint or disjoint
Verify: P(A∪B)=P(A)+P(B)


```python
import random

n = 10000

rolls = [random.randint(1, 6) for _ in range(n)]

count_A = rolls.count(1)
count_B = rolls.count(6)

A_union_B=count_A + count_B

probability_A= count_A/n
probability_B= count_B/n

probability_A_union_B= A_union_B/n

print("ADDITION RULE FOR DISJOINT SETS\n")

print("Number of 1s: ", count_A)
print("Number of 6s: ", count_B)

print("Probability of getting 1: ", probability_A)
print("Probability of getting 6: ", probability_B)

print("P(A∪B) = ", probability_A_union_B)
print("p(A) + p(B) = ", probability_A + probability_B)

print("P(A U B) = P(A) + P(B) : ", probability_A_union_B == (probability_A + probability_B))
```

5.	Addition rule for overlapping events
For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}
Verify: P(A∪B)=P(A)+P(B)−P(A∩B)


```python
A = {2, 4, 6}
B = {4, 5, 6}

intersection = A & B

union = A | B

P_A = len(A) / 6
P_B = len(B) / 6
P_intersection = len(intersection) / 6
P_union = len(union) / 6

print("A intersection B =", intersection)
print("A union B =", union)

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A intersection B) =", P_intersection)
print("P(A union B) =", P_union)

print("P(A) + P(B) - P(A intersection B) =", P_A + P_B - P_intersection)
```

6.	Multiplication rule for independent events
A coin is tossed and a die is rolled 10,000 times.
Define:
•	obtaining Head
•	obtaining an even number
Verify: P(A∩B)=P(A)P(B)


```python
import random

n = 10000

heads = 0
even = 0
both = 0

for _ in range(n):


    coin = random.choice(["Head", "Tail"])
    die = random.randint(1, 6)

    if coin == "Head":
        heads += 1

    if die % 2 == 0:
        even += 1

    if coin == "Head" and die % 2 == 0:
        both += 1

P_A = heads / n
P_B = even / n
P_A_and_B = both / n

print("Number of Heads:", heads)
print("Number of Even Rolls:", even)
print("Number of Head and Even:", both)

print("\nP(A ∩ B) =", P_A_and_B)
print("P(A) × P(B) =", P_A * P_B)
```
